```python
def freq_table(series):
 fa = series.value_counts()
 fr = series.value_counts(normalize=True).mul(100).round(1)
 faa = fa.cumsum()
 fra = fr.cumsum()
 return pd.DataFrame({'FA':fa,'FR(%)':fr,'FAA':faa,'FRA(%)':fra})
for col in df.select_dtypes(include='object').columns:
 print(f'\n--- {col} ---')
 display(freq_table(df[col]))
```

```python
quant = df.select_dtypes(include='number')
Q1, Q3 = quant.quantile(0.25), quant.quantile(0.75)
AIQ = Q3 - Q1
outliers = ((quant < Q1 - 1.5*AIQ) | (quant > Q3 + 1.5*AIQ)).sum()
outliers.rename('N.º outliers').to_frame()
```

```python
import numpy as np
x, y = df['Salario'], df['Prestacao_Casa']
m, b = np.polyfit(x, y, 1)
df.plot(kind='scatter', x='Salario', y='Prestacao_Casa')
plt.plot(x, m*x + b, color='red')
plt.text(x.min(), y.max(), f'y = {m:.2f}x + {b:.2f}', color='red')
plt.show()
```

```python
from scipy import stats
r, p_pearson = stats.pearsonr(df['Salario'], df['Prestacao_Casa'])
rho, p_spear = stats.spearmanr(df['Salario'], df['Prestacao_Casa'])
print(f'Pearson r = {r:.3f} (p={p_pearson:.4f})')
print(f'Spearman ρ = {rho:.3f} (p={p_spear:.4f})')
print(f'R² = {r**2:.3f}')
```
